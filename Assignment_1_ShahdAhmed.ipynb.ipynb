Question 6: Explain whether sort() and sorted() change the original list.

sort() modifies the original list directly (in-place) and returns None.

sorted() does not modify the original list. It returns a new sorted list.

```python
# Task 2.1
python_students = {"Ahmed", "Sara", "Omar", "Mona"}
sql_students = {"Sara", "Mona", "Ali", "Youssef"}

both_courses = python_students.intersection(sql_students)
either_course = python_students.union(sql_students)
exactly_one = python_students.symmetric_difference(sql_students)

print("Both courses:", sorted(both_courses))
print("Either course:", sorted(either_course))
print("Exactly one course:", sorted(exactly_one))
```

```python
# Task 4.1
def calculate_average(scores):
    if len(scores) == 0:
        return 0
    return sum(scores) / len(scores)
```

```python
# Task 4.2
def get_result(score, passing_score=60):
    if score >= passing_score:
        return "Pass"
    return "Fail"
```

```python
# Task 4.3
def summarize_student(name, scores):
    average = calculate_average(scores)
    result = get_result(average)
    return {
        "name": name,
        "average_score": average,
        "result": result
    }
```

```python
# Task 4.4
print("Empty list:", calculate_average([]))
print("Average exactly 60:", get_result(calculate_average([50, 70])))
print("Custom passing score:", get_result(65, 70))
```

```python
# Task 4.5
student1 = summarize_student("Ahmed", [80, 90, 70])
student2 = summarize_student(name="Sara", scores=[50, 60, 70])
print(student1)
print(student2)
```

```python
# Task 4.6
def count_results(summaries):
    results = {"Pass": 0, "Fail": 0}

    for student in summaries:
        results[student["result"]] += 1

    return results
summaries = [
    summarize_student("Ahmed", [80, 90, 70]),
    summarize_student("Sara", [50, 60, 70]),
    summarize_student("Omar", [40, 50, 55])
]

print(count_results(summaries))
print(count_results([]))
```

Task 6.1 – Why is a set suitable for removing duplicates?
A set is suitable for removing duplicates because it stores only unique values. If we convert a list into a set, repeated values will appear only once. However, the original ordering is not preserved because sets are unordered.
Task 6.2 – Difference between print() and return
print() displays a value on the screen, but it does not send the value back to the caller.

return sends a value back from a function, so it can be stored in a variable or used in another operation.
Task 6.3 – One limitation of lambda expressions
Lambda expressions are limited to a single expression, so they are not suitable for complex operations or multiple statements. For more complex logic, we can use functions defined with def.

```python
# Task 7.1
def clean_customers(records):
    cleaned = []
    seen_ids = set()

    for customer in records:
        if customer["id"] not in seen_ids:
            cleaned.append({
                "id": customer["id"],
                "name": customer["name"].strip(),
                "city": customer["city"].title()
            })
            seen_ids.add(customer["id"])

    return cleaned
```

```python
# Task 7.2
seen_ids = set()
unique_customers = []

for customer in cleaned_customers:
    if customer["id"] not in seen_ids:
        unique_customers.append(customer)
        seen_ids.add(customer["id"])

print(unique_customers)
```

```python
# Task 7.3
city_counts = {}

for customer in cleaned_customers:
    city = customer["city"]
    city_counts[city] = city_counts.get(city, 0) + 1

print(city_counts)
```

```python
# Task 7.4
customers = [
    {"id": 101, "name": " Ahmed ", "city": "cairo"},
    {"id": 102, "name": "Sara", "city": "GIZA"},
    {"id": 101, "name": "Ahmed", "city": "Cairo"},
    {"id": 103, "name": " Omar", "city": "cairo"},
    {"id": 104, "name": "Mona ", "city": "giza"}
]

cleaned_customers = clean_customers(customers)

print("Original data:")
print(customers)

print("Cleaned data:")
print(cleaned_customers)

print("City counts:")
print(city_counts)

print("Empty list test:")
print(clean_customers([]))
```

```python
    # Task 8.1
def summarize_orders(orders):
    totals = {}

    for order in orders:
        if order["status"] == "Completed":
            customer = order["customer"]
            totals[customer] = totals.get(customer, 0) + order["amount"]

    return totals
```

```python
# Task 8.2
orders = [
    {"customer": "Ahmed", "amount": 1200, "status": "Completed"},
    {"customer": "Sara", "amount": 850, "status": "Completed"},
    {"customer": "Ahmed", "amount": 400, "status": "Cancelled"},
    {"customer": "Omar", "amount": 600, "status": "Completed"},
    {"customer": "Sara", "amount": 300, "status": "Completed"}
]

totals = summarize_orders(orders)

print(totals)
print(summarize_orders([]))
```

```python
# Task 8.3
sorted_totals = sorted(
    totals.items(),
    key=lambda item: item[1],
    reverse=True
)

print(sorted_totals)
```

```python
# Task 8.4
highest_customer = max(totals, key=totals.get)
total_revenue = sum(totals.values())

print("Highest customer:", highest_customer)
print("Highest order total:", totals[highest_customer])
print("Overall completed revenue:", total_revenue)
```

```python
# Task 8.5
def get_customers_above(totals, minimum=1000):
    return sorted([
        customer for customer, amount in totals.items()
        if amount > minimum
    ])
print(get_customers_above(totals))
print(get_customers_above(totals, 500))
```

**Task 9.1 – Assigning a New Value to an Existing Dictionary Key**
When you assign a new value to a dictionary key that already exists, the old value is replaced with the new value. The key is not duplicated.


**Task 9.2 – Difference Between remove() and discard() for Sets**
Both methods remove an element from a set.
- remove() raises a KeyError if the element does not exist.
- discard() does not raise an error if the element does not exist.


**Task 9.3 – Modifying a List While Iterating**
You should avoid modifying a list while iterating over it because removing or adding elements can change their positions and cause some elements to be skipped or processed incorrectly.


**Task 9.4 – What Do map() and filter() Return in Python 3?**
In Python 3, both map() and filter() return iterators, not lists.
You can convert their results into lists using list().
